You're designing a simple LED indicator circuit. Your job is to:

### Circuit Requirements:

```python
# Challenge setup and helper functions
import sys
sys.path.append('../../../..')  # Add project root to path

from circuit_sim import Circuit, SimulationEngine
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
import plotly.graph_objects as go
import numpy as np
import time

# Challenge parameters
SUPPLY_VOLTAGE = 5.0  # V
LED_FORWARD_VOLTAGE = 2.1  # V
TARGET_CURRENT = 0.020  # A (20mA)
MAX_SAFE_CURRENT = 0.022  # A (22mA safety limit)
LED_MAX_CURRENT = 0.025  # A (25mA absolute max)

print("⚡ LED Circuit Design Challenge")
print("="*40)
print(f"Supply Voltage: {SUPPLY_VOLTAGE}V")
print(f"LED Forward Voltage: {LED_FORWARD_VOLTAGE}V")
print(f"Target Current: {TARGET_CURRENT*1000}mA")
print(f"Safety Limit: {MAX_SAFE_CURRENT*1000}mA")
print("\n🤔 What resistor value do you need?")
```

## Step 1: Calculate the Required Resistor

**Think it through**:
1. What voltage appears across the resistor?
2. What current should flow through the resistor?
3. Use Ohm's law to find the resistance

Enter your calculation below:

```python
# Resistor calculation interface
def create_resistor_calculator():
    # Input widgets
    voltage_input = widgets.FloatText(
        value=0.0,
        description='Voltage across resistor (V):',
        style={'description_width': '200px'}
    )
    
    current_input = widgets.FloatText(
        value=0.020,
        description='Target current (A):',
        style={'description_width': '200px'}
    )
    
    resistance_input = widgets.FloatText(
        value=0.0,
        description='Calculated resistance (Ω):',
        style={'description_width': '200px'}
    )
    
    # Check calculation button
    check_button = widgets.Button(
        description='Check My Calculation',
        button_style='primary',
        layout=widgets.Layout(width='200px')
    )
    
    # Feedback area
    feedback_area = widgets.HTML()
    
    # Hint button
    hint_button = widgets.Button(
        description='Need a Hint?',
        button_style='warning',
        layout=widgets.Layout(width='150px')
    )
    
    hints_shown = {'count': 0}
    
    def check_calculation(b):
        v_resistor = voltage_input.value
        i_target = current_input.value
        r_calculated = resistance_input.value
        
        # Correct values
        v_correct = SUPPLY_VOLTAGE - LED_FORWARD_VOLTAGE  # 2.9V
        r_correct = v_correct / TARGET_CURRENT  # 145Ω
        
        # Check each value
        voltage_ok = abs(v_resistor - v_correct) < 0.1
        current_ok = abs(i_target - TARGET_CURRENT) < 0.001
        resistance_ok = abs(r_calculated - r_correct) < 10
        
        if voltage_ok and current_ok and resistance_ok:
            # Correct!
            feedback_area.value = f"""
            <div style="background: #e8f5e8; padding: 15px; border-radius: 10px; border: 2px solid green;">
                <h4>🎉 Excellent calculation!</h4>
                <p><strong>Your Answer:</strong> {r_calculated:.0f}Ω</p>
                <p><strong>Correct Answer:</strong> {r_correct:.0f}Ω</p>
                <p><strong>Voltage across resistor:</strong> {v_correct:.1f}V</p>
                <p><strong>Logic:</strong> Supply (5V) - LED (2.1V) = {v_correct:.1f}V across resistor</p>
                <p><strong>Ohm's Law:</strong> R = V/I = {v_correct:.1f}V / {TARGET_CURRENT*1000:.0f}mA = {r_correct:.0f}Ω</p>
                <p>✅ Ready to build and test your circuit!</p>
            </div>
            """
        else:
            # Incorrect - provide specific feedback
            errors = []
            if not voltage_ok:
                errors.append(f"Voltage across resistor: Expected {v_correct:.1f}V, got {v_resistor:.1f}V")
            if not current_ok:
                errors.append(f"Target current: Expected {TARGET_CURRENT*1000:.0f}mA, got {i_target*1000:.1f}mA")
            if not resistance_ok:
                errors.append(f"Resistance calculation: Expected ~{r_correct:.0f}Ω, got {r_calculated:.1f}Ω")
            
            feedback_area.value = f"""
            <div style="background: #fff5f5; padding: 15px; border-radius: 10px; border: 2px solid orange;">
                <h4>🤔 Not quite right - let's fix this!</h4>
                <p><strong>Issues found:</strong></p>
                <ul>{''.join(f'<li>{error}</li>' for error in errors)}</ul>
                <p>💡 <strong>Remember:</strong> The resistor and LED share the supply voltage!</p>
            </div>
            """
    
    def show_hint(b):
        hints_shown['count'] += 1
        
        if hints_shown['count'] == 1:
            hint_text = "💡 <strong>Hint 1:</strong> The 5V supply voltage is divided between the LED (2.1V) and the resistor. What voltage is left for the resistor?"
        elif hints_shown['count'] == 2:
            hint_text = "🔍 <strong>Hint 2:</strong> Voltage across resistor = 5V - 2.1V = 2.9V. Now use Ohm's law: R = V/I"
        else:
            hint_text = "📖 <strong>Solution:</strong> R = 2.9V / 0.020A = 145Ω. The resistor drops 2.9V to limit current to 20mA."
        
        feedback_area.value = f"""
        <div style="background: #fffacd; padding: 10px; border-radius: 8px; margin: 5px 0;">
            {hint_text}
        </div>
        """
    
    check_button.on_click(check_calculation)
    hint_button.on_click(show_hint)
    
    # Display the interface
    instruction_html = HTML("""
    <div style="background: #f0f8ff; padding: 15px; border-radius: 10px; margin: 10px 0;">
        <h4>🧮 Calculation Workspace</h4>
        <p>Work through the calculation step by step:</p>
        <ol>
            <li>What voltage appears across the resistor?</li>
            <li>What current do you want through the circuit?</li>
            <li>Use Ohm's Law: R = V/I</li>
        </ol>
    </div>
    """)
    
    display(instruction_html)
    display(voltage_input)
    display(current_input)
    display(resistance_input)
    display(widgets.HBox([check_button, hint_button]))
    display(feedback_area)
    
    return resistance_input

# Create the calculator interface
resistance_widget = create_resistor_calculator()
```

## Step 2: Build and Test Your Circuit

Once you've calculated the resistor value, let's build the actual circuit and verify it works:

```python
# Circuit builder and tester
def create_led_circuit_tester():
    # Get the calculated resistance
    r_value = resistance_widget.value if resistance_widget.value > 0 else 145
    
    print(f"🔧 Building LED circuit with {r_value:.0f}Ω resistor...")
    
    # Build the circuit
    led_circuit = Circuit("LED Current Limiter")
    led_circuit.add_voltage_source("V_supply", 1, 0, "5V")
    led_circuit.add_resistor("R_limit", 1, 2, str(r_value))
    # Model LED as voltage source (simplified model)
    led_circuit.add_voltage_source("LED", 2, 0, "2.1V")
    
    print("✅ Circuit built with components:")
    print(f"   - 5V Power Supply")
    print(f"   - {r_value:.0f}Ω Current Limiting Resistor")
    print(f"   - LED (2.1V forward voltage)")
    
    # Simulate the circuit
    engine = SimulationEngine()
    try:
        results = engine.simulate_dc(led_circuit)
        
        # Extract results
        voltage_at_resistor = results.voltage(1)[0] - results.voltage(2)[0]
        current_through_circuit = abs(results.current("R_limit")[0])
        power_in_resistor = voltage_at_resistor * current_through_circuit
        power_in_led = LED_FORWARD_VOLTAGE * current_through_circuit
        
        print("\n📊 Simulation Results:")
        print("="*40)
        
        # Safety check
        current_safe = current_through_circuit <= MAX_SAFE_CURRENT
        current_status = "✅ SAFE" if current_safe else "⚠️ TOO HIGH"
        
        results_table = f"""
        <div style="background: {'#e8f5e8' if current_safe else '#fff5f5'}; padding: 20px; border-radius: 10px; border: 3px solid {'green' if current_safe else 'red'};">
            <h3>🔬 Circuit Analysis Results</h3>
            <table style="width: 100%; font-size: 16px; font-family: monospace;">
                <tr style="background: #f0f8f0;">
                    <td><strong>Parameter</strong></td>
                    <td><strong>Calculated</strong></td>
                    <td><strong>Simulated</strong></td>
                    <td><strong>Status</strong></td>
                </tr>
                <tr>
                    <td>LED Current</td>
                    <td>{TARGET_CURRENT*1000:.1f}mA</td>
                    <td><strong>{current_through_circuit*1000:.1f}mA</strong></td>
                    <td>{current_status}</td>
                </tr>
                <tr style="background: #f8f8f8;">
                    <td>Voltage across resistor</td>
                    <td>{SUPPLY_VOLTAGE - LED_FORWARD_VOLTAGE:.1f}V</td>
                    <td><strong>{voltage_at_resistor:.2f}V</strong></td>
                    <td>{'✅ Match' if abs(voltage_at_resistor - (SUPPLY_VOLTAGE - LED_FORWARD_VOLTAGE)) < 0.1 else '❌ Off'}</td>
                </tr>
                <tr>
                    <td>Power in resistor</td>
                    <td>{(SUPPLY_VOLTAGE - LED_FORWARD_VOLTAGE) * TARGET_CURRENT * 1000:.1f}mW</td>
                    <td><strong>{power_in_resistor*1000:.1f}mW</strong></td>
                    <td>{'✅ Good' if power_in_resistor < 0.25 else '⚠️ Hot'}</td>
                </tr>
                <tr style="background: #f8f8f8;">
                    <td>Power in LED</td>
                    <td>{LED_FORWARD_VOLTAGE * TARGET_CURRENT * 1000:.1f}mW</td>
                    <td><strong>{power_in_led*1000:.1f}mW</strong></td>
                    <td>{'✅ Safe' if power_in_led < 0.1 else '⚠️ Check'}</td>
                </tr>
            </table>
        </div>
        """
        
        display(HTML(results_table))
        
        # Design assessment
        if current_safe and abs(current_through_circuit - TARGET_CURRENT) < 0.002:
            assessment_html = """
            <div style="background: #e8f5e8; padding: 15px; border-radius: 10px; margin: 10px 0;">
                <h4>🎉 EXCELLENT DESIGN!</h4>
                <p>✅ Current is within safe limits</p>
                <p>✅ Very close to target current (20mA)</p>
                <p>✅ Power dissipation is reasonable</p>
                <p>🏆 <strong>You've successfully designed a safe LED circuit!</strong></p>
            </div>
            """
        else:
            issues = []
            if not current_safe:
                issues.append("Current exceeds safety limit")
            if abs(current_through_circuit - TARGET_CURRENT) >= 0.002:
                issues.append("Current differs significantly from target")
            
            assessment_html = f"""
            <div style="background: #fff5f5; padding: 15px; border-radius: 10px; margin: 10px 0;">
                <h4>🔧 Design Needs Adjustment</h4>
                <p><strong>Issues:</strong> {', '.join(issues)}</p>
                <p>💡 <strong>Suggestion:</strong> Recalculate the resistor value or check your circuit connections</p>
            </div>
            """
        
        display(HTML(assessment_html))
        
        return True, current_through_circuit, voltage_at_resistor
        
    except Exception as e:
        error_html = f"""
        <div style="background: #fff5f5; padding: 15px; border-radius: 10px; border: 2px solid red;">
            <h4>❌ Simulation Error</h4>
            <p><strong>Error:</strong> {str(e)}</p>
            <p>This might be due to an invalid resistor value or circuit configuration.</p>
            <p>Try recalculating the resistor value.</p>
        </div>
        """
        display(HTML(error_html))
        return False, None, None

# Test the LED circuit
print("🧪 Testing your LED circuit design...\n")
success, current_result, voltage_result = create_led_circuit_tester()
```

## Step 3: Component Selection Reality Check

In the real world, you can't buy a resistor with any arbitrary value. Let's see what standard resistor values are available:

```python
# Standard resistor values and selection
def standard_resistor_selection():
    # E12 standard values (10% tolerance)
    e12_base = [10, 12, 15, 18, 22, 27, 33, 39, 47, 56, 68, 82]
    standard_values = []
    
    # Generate standard values from 100Ω to 1kΩ
    for decade in [10, 100]:
        for base in e12_base:
            standard_values.append(base * decade)
    
    # Calculate ideal resistor
    ideal_r = (SUPPLY_VOLTAGE - LED_FORWARD_VOLTAGE) / TARGET_CURRENT
    
    # Find closest standard values
    closest_values = sorted(standard_values, key=lambda x: abs(x - ideal_r))[:3]
    
    print(f"\n🎯 Resistor Selection Analysis")
    print("="*45)
    print(f"Ideal calculated value: {ideal_r:.1f}Ω")
    print(f"\nClosest standard values (E12 series):")
    
    results_data = []
    
    for i, r_std in enumerate(closest_values):
        # Calculate resulting current with this resistor
        actual_current = (SUPPLY_VOLTAGE - LED_FORWARD_VOLTAGE) / r_std
        current_error = abs(actual_current - TARGET_CURRENT) / TARGET_CURRENT * 100
        is_safe = actual_current <= MAX_SAFE_CURRENT
        
        status = "✅ GOOD" if is_safe and current_error < 10 else "⚠️ CHECK" if is_safe else "❌ UNSAFE"
        
        print(f"  {i+1}. {r_std:3d}Ω → {actual_current*1000:5.1f}mA ({current_error:4.1f}% error) {status}")
        
        results_data.append({
            'resistance': r_std,
            'current': actual_current,
            'error': current_error,
            'safe': is_safe
        })
    
    # Create comparison chart
    fig = go.Figure()
    
    resistances = [d['resistance'] for d in results_data]
    currents = [d['current'] * 1000 for d in results_data]  # Convert to mA
    colors = ['green' if d['safe'] and d['error'] < 10 else 'orange' if d['safe'] else 'red' for d in results_data]
    
    fig.add_trace(go.Bar(
        x=[f"{r}Ω" for r in resistances],
        y=currents,
        marker_color=colors,
        text=[f"{c:.1f}mA" for c in currents],
        textposition='auto',
        name='LED Current'
    ))
    
    # Add target line
    fig.add_hline(y=TARGET_CURRENT*1000, line_dash="dash", line_color="blue", 
                  annotation_text=f"Target: {TARGET_CURRENT*1000}mA")
    
    # Add safety limit line
    fig.add_hline(y=MAX_SAFE_CURRENT*1000, line_dash="dash", line_color="red",
                  annotation_text=f"Safety Limit: {MAX_SAFE_CURRENT*1000}mA")
    
    fig.update_layout(
        title={
            'text': '📊 Standard Resistor Values Comparison',
            'x': 0.5,
            'xanchor': 'center'
        },
        xaxis_title='Standard Resistor Value',
        yaxis_title='LED Current (mA)',
        showlegend=False,
        height=400,
        template='plotly_white'
    )
    
    fig.show()
    
    # Recommendation
    best_choice = min(results_data, key=lambda x: x['error'] if x['safe'] else float('inf'))
    
    recommendation_html = f"""
    <div style="background: #e8f5e8; padding: 20px; border-radius: 10px; border: 2px solid green; margin: 10px 0;">
        <h4>🎯 Professional Recommendation</h4>
        <p><strong>Best choice:</strong> {best_choice['resistance']}Ω standard resistor</p>
        <p><strong>Resulting current:</strong> {best_choice['current']*1000:.1f}mA</p>
        <p><strong>Error from target:</strong> {best_choice['error']:.1f}%</p>
        <p>✅ This is close enough for most applications and uses a readily available component!</p>
        
        <h5>🛒 What to buy:</h5>
        <ul>
            <li><strong>Resistor:</strong> {best_choice['resistance']}Ω, ¼W, 5% or 10% tolerance</li>
            <li><strong>LED:</strong> Standard 5mm red LED (2V-2.2V forward voltage)</li>
            <li><strong>Power supply:</strong> 5V DC adapter</li>
        </ul>
    </div>
    """
    
    display(HTML(recommendation_html))

# Run the standard resistor analysis
if success:
    standard_resistor_selection()
else:
    print("⚠️ Skipping resistor selection due to simulation error")
```
