## Confidence Self-Assessment

Rate your confidence level (1-5 stars) for each concept:

```python
# Self-assessment interface
def create_self_assessment():
    # Define concepts to assess
    concepts = {
        'dc_concept': 'Understanding what DC analysis calculates',
        'voltage_measurement': 'Predicting voltages at circuit connection points',
        'ohms_law': 'Using Ohm\'s law (V=IR) for circuit calculations',
        'current_limiting': 'Designing current-limiting resistors for LEDs',
        'power_calculation': 'Calculating power dissipation in resistors',
        'safety_margins': 'Applying safety factors in real designs',
        'standard_components': 'Selecting standard resistor values'
    }
    
    # Create assessment widgets
    assessments = {}
    
    assessment_html = HTML("""
    <div style="background: #f8f9fa; padding: 15px; border-radius: 10px; margin: 10px 0;">
        <h4>🎯 Rate Your Confidence (1-5 stars)</h4>
        <p><strong>1 star</strong> = "I'm confused" | <strong>3 stars</strong> = "I understand" | <strong>5 stars</strong> = "I could teach this"</p>
    </div>
    """)
    display(assessment_html)
    
    for concept_key, description in concepts.items():
        assessments[concept_key] = widgets.IntSlider(
            value=3,
            min=1,
            max=5,
            step=1,
            description=description,
            style={'description_width': '350px'},
            layout={'width': '500px'}
        )
        display(assessments[concept_key])
    
    # Additional feedback questions
    difficulty_rating = widgets.RadioButtons(
        options=[
            ('Too Easy - I was bored', 'too_easy'),
            ('Just Right - Good challenge', 'just_right'),
            ('Too Hard - I struggled a lot', 'too_hard')
        ],
        description='Overall difficulty:',
        style={'description_width': '150px'}
    )
    
    time_rating = widgets.RadioButtons(
        options=[
            ('Too Fast - I felt rushed', 'too_fast'),
            ('Good Pace - Comfortable timing', 'good_pace'),
            ('Too Slow - I wanted to move faster', 'too_slow')
        ],
        description='Learning pace:',
        style={'description_width': '150px'}
    )
    
    display(HTML("<h4>📝 Additional Feedback:</h4>"))
    display(difficulty_rating)
    display(time_rating)
    
    # Analysis button
    analyze_button = widgets.Button(
        description='Get My Learning Recommendation',
        button_style='success',
        layout=widgets.Layout(width='300px', height='50px')
    )
    
    results_area = widgets.HTML()
    
    def analyze_assessment(b):
        # Calculate average confidence
        confidence_scores = [widget.value for widget in assessments.values()]
        avg_confidence = np.mean(confidence_scores)
        min_confidence = min(confidence_scores)
        
        # Identify weak areas
        weak_concepts = [desc for (key, desc), score in zip(concepts.items(), confidence_scores) if score < 3]
        strong_concepts = [desc for (key, desc), score in zip(concepts.items(), confidence_scores) if score >= 4]
        
        # Generate personalized recommendation
        if avg_confidence >= 4.0:
            # High confidence - ready to advance
            recommendation = "advance"
            color = "green"
            title = "🚀 Ready to Advance!"
            message = f"Excellent work! Your average confidence is {avg_confidence:.1f}/5. You've mastered the fundamentals of DC analysis."
            next_step = "**Next Module**: Voltage Dividers and More Complex Circuits"
            
        elif avg_confidence >= 3.0 and min_confidence >= 2:
            # Medium confidence - can advance with review
            recommendation = "advance_with_review"
            color = "blue"
            title = "📈 Good Progress - Ready with Review"
            message = f"Good foundation! Your average confidence is {avg_confidence:.1f}/5. Consider reviewing weak areas first."
            next_step = "**Recommended**: Quick review exercises, then advance to Module 1.2"
            
        else:
            # Low confidence - needs more practice
            recommendation = "more_practice"
            color = "orange"
            title = "🔄 Let's Reinforce These Concepts"
            message = f"Your confidence is {avg_confidence:.1f}/5. Let's strengthen your foundation with more practice."
            next_step = "**Recommended**: Additional practice exercises on weak areas"
        
        # Create detailed feedback
        feedback_html = f"""
        <div style="background: #f0f8ff; padding: 20px; border-radius: 15px; border: 3px solid {color}; margin: 15px 0;">
            <h3>{title}</h3>
            <p style="font-size: 16px;">{message}</p>
            
            <h4>📊 Your Confidence Profile:</h4>
            <ul>
                <li><strong>Average Confidence:</strong> {avg_confidence:.1f}/5 stars</li>
                <li><strong>Strongest Areas:</strong> {len(strong_concepts)} concepts</li>
                <li><strong>Areas for Growth:</strong> {len(weak_concepts)} concepts</li>
            </ul>
        """
        
        if strong_concepts:
            feedback_html += f"""
            <h4>✅ <strong>Your Strengths:</strong></h4>
            <ul>
                {''.join(f'<li>{concept}</li>' for concept in strong_concepts[:3])}
            </ul>
            """
        
        if weak_concepts:
            feedback_html += f"""
            <h4>📚 <strong>Focus Areas:</strong></h4>
            <ul>
                {''.join(f'<li>{concept}</li>' for concept in weak_concepts)}
            </ul>
            """
        
        # Add pace and difficulty feedback
        difficulty_feedback = {
            'too_easy': '⬆️ We\'ll increase the challenge level',
            'just_right': '✅ Perfect difficulty level maintained',
            'too_hard': '⬇️ We\'ll provide more scaffolding and support'
        }
        
        pace_feedback = {
            'too_fast': '🐌 We\'ll slow down and provide more time',
            'good_pace': '⏱️ Maintaining current pacing',
            'too_slow': '⚡ We\'ll pick up the pace'
        }
        
        feedback_html += f"""
        <h4>⚙️ <strong>Personalization Adjustments:</strong></h4>
        <ul>
            <li><strong>Difficulty:</strong> {difficulty_feedback.get(difficulty_rating.value, 'No adjustment')}</li>
            <li><strong>Pace:</strong> {pace_feedback.get(time_rating.value, 'No adjustment')}</li>
        </ul>
        
        <div style="background: rgba(255,255,255,0.8); padding: 15px; border-radius: 8px; margin: 10px 0;">
            <h4>🎯 <strong>Recommended Next Step:</strong></h4>
            <p style="font-size: 18px; font-weight: bold;">{next_step}</p>
        </div>
        """
        
        if recommendation == "advance":
            feedback_html += """
            <button onclick="window.open('../module_1.2_voltage_dividers/explain_voltage_division.ipynb', '_blank')" 
                    style="background: green; color: white; padding: 15px 30px; border: none; border-radius: 8px; font-size: 16px; cursor: pointer; margin: 10px 5px;">
                🚀 Continue to Module 1.2
            </button>
            """
        elif recommendation == "advance_with_review":
            feedback_html += """
            <button onclick="alert('Review exercises coming soon!')" 
                    style="background: blue; color: white; padding: 15px 30px; border: none; border-radius: 8px; font-size: 16px; cursor: pointer; margin: 10px 5px;">
                📚 Quick Review First
            </button>
            <button onclick="window.open('../module_1.2_voltage_dividers/explain_voltage_division.ipynb', '_blank')" 
                    style="background: green; color: white; padding: 15px 30px; border: none; border-radius: 8px; font-size: 16px; cursor: pointer; margin: 10px 5px;">
                ➡️ Continue to Module 1.2
            </button>
            """
        else:
            feedback_html += """
            <button onclick="alert('Additional practice exercises coming soon!')" 
                    style="background: orange; color: white; padding: 15px 30px; border: none; border-radius: 8px; font-size: 16px; cursor: pointer; margin: 10px 5px;">
                🔄 More Practice Exercises
            </button>
            <button onclick="window.location.reload()" 
                    style="background: gray; color: white; padding: 15px 30px; border: none; border-radius: 8px; font-size: 16px; cursor: pointer; margin: 10px 5px;">
                🔁 Retake Module 1.1
            </button>
            """
        
        feedback_html += "</div>"
        
        results_area.value = feedback_html
        
        # Save assessment data (would be stored in real implementation)
        assessment_data = {
            'module': '1.1_dc_basics',
            'avg_confidence': avg_confidence,
            'min_confidence': min_confidence,
            'weak_concepts': weak_concepts,
            'difficulty_rating': difficulty_rating.value,
            'time_rating': time_rating.value,
            'recommendation': recommendation,
            'timestamp': time.time()
        }
        
        print(f"\n✅ Assessment saved: {recommendation} pathway recommended")
    
    analyze_button.on_click(analyze_assessment)
    
    display(HTML("<div style='margin: 20px 0;'></div>"))
    display(analyze_button)
    display(results_area)
    
    return assessments

# Create the self-assessment interface
assessment_widgets = create_self_assessment()
```

## 📈 Learning Analytics Dashboard

Here's how you're progressing through the circuit simulation learning path:

```python
# Learning progress visualization
import plotly.graph_objects as go
from plotly.subplots import make_subplots

def create_progress_dashboard():
    # Simulated progress data (would come from user's actual performance)
    modules_data = {
        'Module 1.1\nDC Basics': {'completed': True, 'score': 85, 'time_spent': 15},
        'Module 1.2\nVoltage Dividers': {'completed': False, 'score': 0, 'time_spent': 0},
        'Module 1.3\nCurrent Limiting': {'completed': False, 'score': 0, 'time_spent': 0},
        'Module 1.4\nComplex Networks': {'completed': False, 'score': 0, 'time_spent': 0},
    }
    
    # Create subplots
    fig = make_subplots(
        rows=1, cols=2,
        subplot_titles=('Learning Progress', 'Time Spent per Module'),
        specs=[[{'type': 'bar'}, {'type': 'bar'}]]
    )
    
    # Progress chart
    modules = list(modules_data.keys())
    scores = [data['score'] for data in modules_data.values()]
    colors = ['green' if data['completed'] else 'lightgray' for data in modules_data.values()]
    
    fig.add_trace(
        go.Bar(
            x=modules,
            y=scores,
            marker_color=colors,
            text=[f"{s}%" if s > 0 else "Not Started" for s in scores],
            textposition='auto',
            name='Completion Score'
        ),
        row=1, col=1
    )
    
    # Time spent chart
    times = [data['time_spent'] for data in modules_data.values()]
    
    fig.add_trace(
        go.Bar(
            x=modules,
            y=times,
            marker_color='blue',
            text=[f"{t} min" if t > 0 else "0 min" for t in times],
            textposition='auto',
            name='Time Spent'
        ),
        row=1, col=2
    )
    
    # Update layout
    fig.update_layout(
        title={
            'text': '📊 Your Learning Progress Dashboard',
            'x': 0.5,
            'xanchor': 'center',
            'font': {'size': 18}
        },
        showlegend=False,
        height=400
    )
    
    fig.update_yaxes(title_text="Completion Score (%)", row=1, col=1)
    fig.update_yaxes(title_text="Time (minutes)", row=1, col=2)
    
    fig.show()
    
    # Summary statistics
    total_modules = len(modules_data)
    completed_modules = sum(1 for data in modules_data.values() if data['completed'])
    total_time = sum(data['time_spent'] for data in modules_data.values())
    avg_score = np.mean([data['score'] for data in modules_data.values() if data['score'] > 0]) if completed_modules > 0 else 0
    
    summary_html = f"""
    <div style="background: #f8f9fa; padding: 20px; border-radius: 10px; margin: 10px 0;">
        <h4>📈 Learning Statistics</h4>
        <div style="display: grid; grid-template-columns: repeat(2, 1fr); gap: 20px; font-family: monospace;">
            <div>
                <strong>Overall Progress:</strong><br>
                Modules completed: {completed_modules}/{total_modules}<br>
                Average score: {avg_score:.1f}%<br>
                Total time invested: {total_time} minutes
            </div>
            <div>
                <strong>Track 1 (DC Analysis):</strong><br>
                Progress: {completed_modules}/4 modules<br>
                Estimated time remaining: {(4-completed_modules)*12} minutes<br>
                On track for mastery: {'✅ Yes' if avg_score >= 70 else '⚠️ Needs improvement'}
            </div>
        </div>
    </div>
    """
    
    display(HTML(summary_html))

# Create the progress dashboard
create_progress_dashboard()
```
