## The Challenge

Look at this simple circuit:

```
    9V Battery
        |
        +---[1000Ω Resistor]---+
        |                      |
        A                    Ground
```

**Question**: What voltage would you measure at point **A** with a voltmeter?

---

````python
# Interactive voltage prediction exercise
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
import time

# Setup the exercise
def create_voltage_prediction_exercise():
    # Question display
    question_html = HTML("""
    <div style="background: #f0f8ff; padding: 20px; border-radius: 10px; margin: 10px 0;">
        <h3>🤔 What voltage will you measure at point A?</h3>
        <div style="font-family: monospace; font-size: 14px; background: white; padding: 10px; border-radius: 5px;">
            9V Battery<br>
            &nbsp;&nbsp;&nbsp;&nbsp;|<br>
            &nbsp;&nbsp;&nbsp;&nbsp;+---[1000Ω]---+<br>
            &nbsp;&nbsp;&nbsp;&nbsp;|&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;|<br>
            &nbsp;&nbsp;&nbsp;&nbsp;A&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Ground
        </div>
    </div>
    """)
    
    # Answer options
    answer_widget = widgets.RadioButtons(
        options=[
            ('3V', '3V'),
            ('4.5V', '4.5V'), 
            ('9V', '9V'),
            ('0V', '0V')
        ],
        description='Your answer:',
        style={'description_width': '100px'}
    )
    
    # Submit button
    submit_button = widgets.Button(
        description='Check Answer',
        button_style='primary',
        layout=widgets.Layout(width='150px')
    )
    
    # Feedback area
    feedback_area = widgets.HTML()
    
    # Track attempts and time
    start_time = time.time()
    attempts = {'count': 0}
    
    def check_answer(b):
        attempts['count'] += 1
        elapsed_time = time.time() - start_time
        
        if answer_widget.value == '9V':
            # Correct answer!
            feedback_area.value = f"""
            <div style="background: #e8f5e8; padding: 15px; border-radius: 10px; border: 2px solid green;">
                <h4>🎉 Excellent! You got it right!</h4>
                <p><strong>Answer: 9V</strong></p>
                <p><strong>Why:</strong> Point A is directly connected to the positive terminal of the battery. 
                Even though current flows through the resistor to ground, the voltage at point A 
                is still the full battery voltage!</p>
                <p><strong>Time:</strong> {elapsed_time:.1f} seconds | <strong>Attempts:</strong> {attempts['count']}</p>
                <p>Think of it like water pressure - the pressure is the same throughout the pipe, 
                even if water is flowing.</p>
                <button onclick="location.href='build_first_circuit.ipynb'" 
                        style="background: green; color: white; padding: 10px 20px; border: none; border-radius: 5px; cursor: pointer;">
                    Continue to Build Exercise →
                </button>
            </div>
            """
            
        else:
            # Incorrect answer - provide helpful hint
            if attempts['count'] == 1:
                hint = "💡 <strong>Hint:</strong> Think about what point A is connected to. Does it matter that current flows through the resistor?"
            elif attempts['count'] == 2:
                hint = "🔍 <strong>Stronger Hint:</strong> Point A is directly connected to the positive terminal of the battery. What's the voltage at the positive terminal?"
            else:
                hint = "📖 <strong>Explanation:</strong> Point A connects directly to the battery's positive terminal (9V). Current flow doesn't change voltage at connection points!"
            
            feedback_area.value = f"""
            <div style="background: #fff5f5; padding: 15px; border-radius: 10px; border: 2px solid orange;">
                <h4>🤔 Not quite right, but that's okay!</h4>
                <p>You selected: <strong>{answer_widget.value}</strong></p>
                <p>{hint}</p>
                <p><strong>Attempts:</strong> {attempts['count']}/3 | Keep thinking! 💪</p>
            </div>
            """
    
    submit_button.on_click(check_answer)
    
    # Display everything
    display(question_html)
    display(answer_widget)
    display(submit_button)
    display(feedback_area)

# Run the exercise
create_voltage_prediction_exercise()
````

---

## Learning Notes

**Key Concept**: Voltage exists at connection points regardless of current flow.

**Analogy**: Like water pressure in pipes - the pressure is the same throughout the pipe, even when water is flowing.

**Next**: Once you get this right, we'll build an actual circuit and verify this with simulation!

---

### What You Just Learned
- Voltage measurements tell you the "electrical pressure" at any point
- Points directly connected to a voltage source have that source's voltage
- Current flow through components doesn't affect voltage at connection points

This is **fundamental** - make sure you understand it before moving on!
